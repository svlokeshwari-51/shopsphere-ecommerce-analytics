## Cleaning Principles

1. Raw CSV files are not modified.
2. Cleaning is performed on working/processed datasets.
3. Missing values are not automatically deleted.
4. Suspicious values are investigated before removal.
5. Date columns are converted to datetime where date-based analysis is required.
6. Business rules are defined before cleaning.
7. Cleaned data is validated against the original data.
8. All important cleaning decisions are documented.

# ShopSphere — Data Quality Summary

## Key Data Quality Findings

| Table | Issue | Business Impact | Decision |
|---|---|---|---|
| Customers | Missing city values | Affects city-level analysis | Retain records; handle missing city explicitly |
| Products | Missing brand values | Affects brand-level analysis | Retain products |
| Order Items | Invalid quantity values | Can distort revenue/profit | Investigate and clean invalid records |
| Marketing | Missing acquisition channel | Affects channel analysis | Handle missing values explicitly |
| Payments | Date/status validation | Affects revenue analysis | Validate before calculating revenue |
| Returns | Return quantity/refund validation | Affects net revenue and profit | Validate before analysis |
| Experiment Sessions | Event consistency | Affects A/B test validity | Validate before statistical analysis |

# Data Cleaning

The raw datasets are preserved unchanged. Cleaning decisions are based on the profiling results and documented business rules.

```python
customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
products_clean = products.copy()
categories_clean = categories.copy()
payments_clean = payments.copy()
returns_clean = returns.copy()
marketing_clean = marketing.copy()
experiment_clean = experiment_sessions.copy()
```

```python
customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"]
)

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"]
)

payments_clean["payment_date"] = pd.to_datetime(
    payments_clean["payment_date"]
)

returns_clean["return_date"] = pd.to_datetime(
    returns_clean["return_date"]
)

marketing_clean["acquisition_date"] = pd.to_datetime(
    marketing_clean["acquisition_date"]
)

experiment_clean["session_date"] = pd.to_datetime(
    experiment_clean["session_date"]
)
```

```python
customers_clean.duplicated().sum()
```

```python
customers_clean[
    customers_clean.duplicated(keep=False)
].sort_values("customer_id")
```

```python
customers_clean = customers_clean.drop_duplicates().copy()
```

```python
print("Remaining exact duplicates:",
      customers_clean.duplicated().sum())

print("Unique customer IDs:",
      customers_clean["customer_id"].nunique())

print("Rows:",
      len(customers_clean))
```

```python
customers_clean["city"].isnull().sum()
```

```python
customers_clean["city"] = customers_clean["city"].fillna("Unknown")
```

```python
customers_clean["city"].isnull().sum()
```

```python
products_clean["brand"].isnull().sum()
```

```python
products_clean["brand"] = products_clean["brand"].fillna("Unknown")
```

```python
products_clean["brand"].isnull().sum()
```

```python
marketing_clean["acquisition_channel"].isnull().sum()
```

```python
marketing_clean["acquisition_channel"] = (
    marketing_clean["acquisition_channel"]
    .fillna("Unknown")
)
```

```python
marketing_clean["acquisition_channel"].isnull().sum()
```

```python
order_items_clean[order_items_clean["quantity"] < 0]
```

```python
order_items_clean = order_items_clean[
    order_items_clean["quantity"] > 0
].copy()
```

```python


order_items_clean["gross_value"] = (
    order_items_clean["quantity"]
    * order_items_clean["unit_price"]
)
```

```python
order_items_clean[
    order_items_clean["discount_amount"]
    > order_items_clean["gross_value"]
]
```

```python
excess_discount_mask = (
    order_items_clean["discount_amount"]
    > order_items_clean["gross_value"]
)

print("Excessive discount rows:",
      excess_discount_mask.sum())

print("Total excessive discount amount:",
      (
          order_items_clean.loc[
              excess_discount_mask,
              "discount_amount"
          ].sum()
      ))
```

```python
excess_discount_mask = (
    order_items_clean["discount_amount"]
    > order_items_clean["gross_value"]
)

print("Affected rows:", excess_discount_mask.sum())
print(
    "Affected gross value:",
    order_items_clean.loc[
        excess_discount_mask, "gross_value"
    ].sum()
)
print(
    "Affected discount amount:",
    order_items_clean.loc[
        excess_discount_mask, "discount_amount"
    ].sum()
)
```

```python
order_items_clean = order_items_clean[
    ~excess_discount_mask
].copy()
```

```python
print(
    "Invalid quantities:",
    (order_items_clean["quantity"] <= 0).sum()
)

print(
    "Discount > gross value:",
    (
        order_items_clean["discount_amount"]
        > order_items_clean["gross_value"]
    ).sum()
)

print("Original rows:", len(order_items))
print("Cleaned rows:", len(order_items_clean))
print("Rows removed:", len(order_items) - len(order_items_clean))
```

```python
order_items_clean["net_item_value"] = (
    order_items_clean["gross_value"]
    - order_items_clean["discount_amount"]
)
```
